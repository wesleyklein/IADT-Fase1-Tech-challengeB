# 03 - Conversão da variável alvo
Este notebook transforma a coluna `diagnosis` em uma variável numérica, etapa necessária para o treinamento dos algoritmos de machine learning.

Aqui adotamos a convenção:
- `M` (maligno) -> `1`
- `B` (benigno) -> `0`

Assim, a variável alvo fica pronta para ser interpretada pelos modelos supervisionados.

```python
import pandas as pd

# Carregar a versão limpa da base, já sem a coluna de identificação.
df = pd.read_csv('../data/breast-cancer-wisconsin-data-clean.csv')

# Converter a coluna de diagnóstico de texto para número.
# M = maligno (1) | B = benigno (0).
# Essa codificação é necessária para a etapa de modelagem.
df['diagnostico'] = df['diagnostico'].map({'M': 1, 'B': 0})
```

```python
# Conferir a distribuição da variável alvo após a codificação.
# Se tudo deu certo, agora veremos 0 e 1 no lugar de B e M.
df['diagnostico'].value_counts()
```

```python
# Salvar a base atualizada com a coluna de diagnóstico numérica.
# A partir daqui, os próximos notebooks já podem trabalhar com a variável alvo pronta.
df.to_csv('../data/breast-cancer-wisconsin-data-clean.csv', index=False)
```

```python
# Reabrir o arquivo salvo para validar se a transformação foi gravada em disco.
df = pd.read_csv('../data/breast-cancer-wisconsin-data-clean.csv')

# Conferir novamente a contagem das classes após salvar o arquivo.
df['diagnostico'].value_counts()
```

### Conclusão da etapa
Ao final deste notebook, a variável `diagnosis` deixa de ser textual e passa a ser numérica, o que é essencial para o treinamento dos modelos.

Essa transformação encerra a preparação básica da variável alvo e deixa a base pronta para análises de correlação, visualização e modelagem.
