# 02 - Limpeza inicial da base
Neste notebook fazemos a **primeira limpeza estrutural** do dataset.

O foco aqui é remover colunas que não ajudam o modelo a aprender padrões clínicos.
Neste caso, a coluna `id` é excluída porque funciona apenas como identificador do exame/paciente e não como característica médica útil para predição.

```python
import pandas as pd

# Carregar a base original.
df = pd.read_csv('../data/breast-cancer-wisconsin-data.csv')

# Remover colunas inválidas ou identificadores que não devem entrar no modelo.
# - 'Unnamed: 32' pode aparecer como artefato vazio da planilha original.
# - 'id' é apenas identificador e não carrega informação clínica relevante.
df = df.drop(columns=['Unnamed: 32', 'id'], errors='ignore')

# Remover colunas completamente vazias, caso existam.
df = df.dropna(axis=1, how='all')

# Converter a coluna de diagnóstico de texto para número.
# M = maligno (1) | B = benigno (0).
# Essa codificação é necessária para a etapa de modelagem.
df['diagnosis'] = df['diagnosis'].map({'M': 1, 'B': 0})

# Renomear colunas para português.
colunas_traduzidas = {
    'diagnosis': 'diagnostico',

    'radius_mean': 'raio_media',
    'texture_mean': 'textura_media',
    'perimeter_mean': 'perimetro_media',
    'area_mean': 'area_media',
    'smoothness_mean': 'suavidade_media',
    'compactness_mean': 'compactacao_media',
    'concavity_mean': 'concavidade_media',
    'concave points_mean': 'pontos_concavos_media',
    'symmetry_mean': 'simetria_media',
    'fractal_dimension_mean': 'dimensao_fractal_media',

    'radius_se': 'raio_erro_padrao',
    'texture_se': 'textura_erro_padrao',
    'perimeter_se': 'perimetro_erro_padrao',
    'area_se': 'area_erro_padrao',
    'smoothness_se': 'suavidade_erro_padrao',
    'compactness_se': 'compactacao_erro_padrao',
    'concavity_se': 'concavidade_erro_padrao',
    'concave points_se': 'pontos_concavos_erro_padrao',
    'symmetry_se': 'simetria_erro_padrao',
    'fractal_dimension_se': 'dimensao_fractal_erro_padrao',

    'radius_worst': 'raio_pior',
    'texture_worst': 'textura_pior',
    'perimeter_worst': 'perimetro_pior',
    'area_worst': 'area_pior',
    'smoothness_worst': 'suavidade_pior',
    'compactness_worst': 'compactacao_pior',
    'concavity_worst': 'concavidade_pior',
    'concave points_worst': 'pontos_concavos_pior',
    'symmetry_worst': 'simetria_pior',
    'fractal_dimension_worst': 'dimensao_fractal_pior'
}

df = df.rename(columns=colunas_traduzidas)

# Salvar nova base tratada com colunas em português.
df.to_csv('../data/breast-cancer-wisconsin-data-clean.csv', index=False)

# Visualizar as primeiras linhas da nova base.
df.head()
```

### Resultado esperado desta etapa
Ao final deste notebook, o arquivo `../data/breast-cancer-wisconsin-data-clean.csv` passa a existir sem a coluna `id`.

Isso prepara a base para a transformação da coluna de diagnóstico em valores numéricos no próximo notebook.
