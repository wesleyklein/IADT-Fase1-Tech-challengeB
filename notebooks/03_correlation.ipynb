# 04 - Análise de correlação
Neste notebook avaliamos como as variáveis numéricas se relacionam entre si e, principalmente, com a variável alvo `diagnosis`.

Essa etapa ajuda a identificar atributos potencialmente mais relevantes para a predição e também mostra quando diferentes colunas carregam informações parecidas.

```python
import pandas as pd

# Carregar a base já limpa e com a variável alvo preparada para análise.
df = pd.read_csv('../data/breast-cancer-wisconsin-data-clean.csv')

import seaborn as sns
import matplotlib.pyplot as plt

# Construir a matriz de correlação entre as variáveis numéricas.
# O mapa de calor ajuda a enxergar relações positivas e negativas
# e a identificar grupos de atributos com comportamento semelhante.
plt.figure(figsize=(12, 10))
sns.heatmap(df.corr(), cmap='coolwarm')
plt.title('Matriz de Correlação')
plt.show()
```

```python
# Ordenar a correlação de cada variável com o diagnóstico.
# Assim identificamos quais atributos parecem mais associados
# à classificação entre benigno e maligno.
correlation = df.corr()['diagnostico'].sort_values(ascending=False)
print(correlation)
```
